```python
bins = [0, 100000, 300000, float('inf')]
labels = ['Low', 'Med', 'High']

df['Incom Band'] = pd.cut(df['Estimated Income'], bins=bins, labels=labels, right=False)
```

```python
# Exmain the distribution of unique cataegories in catergorical columns
categorical_cols = df[['BRId', 'GenderId', 'IAId','Incom Band','Nationality', 'Occupation', 'Fee Structure','Loyalty Classification','Amount of Credit Cards']]

for col in categorical_cols:
    print(f"value count for '{col}':")
    display(df[col].value_counts())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# NUMERICAL ANALYSIS

numerical_cols= ['Estimated Income','Superannuation Savings', 'Credit Card Balance']

# Univeriate analysis and Visualization
plt.figure(figsize = (8,4))
for i,col in enumerate(numerical_cols):
    plt.subplot(3,1,i+1)
    sns.histplot(df[col],kde=True)
    plt.title(col)
    plt.show()
                
```
